#Title

```python
df['title'].values
```

```python
df['title'].isnull().sum()
```

```python
df['title'].unique()
```

```python
# Map gender to title, falling back to 'N/A' if gender is neither male nor female
df['title'] = np.where(
    df['title'].isnull(),
    np.where(
        df['gender'].str.upper().isin(['MALE']),
        'Mr',
        np.where(
            df['gender'].str.upper().isin(['FEMALE']),
            'Mrs',
            'N/A'
        )
    ),
    df['title']
)
```
